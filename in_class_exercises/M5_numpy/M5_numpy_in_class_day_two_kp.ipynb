## Slide examples: axes and reductions (slides 34–36)

```python
# Slide 36: axis=0 collapses the rows -> one result per column
x = np.array([[1, 2, 3], [4, 5, 6]])
print(x)

# adding up by column
print(np.sum(x, axis=0))

# adding up by row
print(np.sum(x, axis=1))   # one result per row

# add everything
print(x.sum())             # no axis: everything
```

## Slide examples: fancy and masked indexing (slides 41–42, 64)

```python
# Slide 41: index with one index, a list of indexes, or an array of indexes
a = np.array([1, 4, 9, 16, 25, 36, 49])
print(a[3])
print(a[[3, 1, 6]])
print(a[np.array([[1, 3], [5, 2]])])   # result has the shape of the index array
```

```python
# Slide 42: select with a boolean mask
c = np.array([100, 200, 300])
mask = np.array([True, False, True])
c[mask]
```

```python
# Relational operations make masks
print(c > 100)
print(c[c > 100])
print(c[(c > 100) & (c / 100 == 2)])   # and: parentheses required
print(c[(c > 100) | (c / 100 == 2)])   # or
```

```python
# Quiz slide 71 (predict first!)
x = np.array([1, 3, 5, 7, 9])
y = np.array([1, 2, 3, 2, 1])
x[y < 2]
```

```python
# Assigning through a mask changes the array in place
m = np.arange(10).reshape(2, 5)
m[(m > 4) & (m % 2 == 0)] = 0
m
```

```python
# Slide 64: np.where -- select, or replace conditionally
m = np.array([[3, 2, 1], [3, 2, 1], [1, 2, 3]])
print(m[np.where(m > 1)])        # same as m[m > 1]
print(np.where(m > 1, m, m + 2))  # keep m where True, else m + 2
```

## Slide examples: broadcasting (slides 46–50)
Rule (slide 50): compare shapes right to left; each pair must be equal, or one of them must be 1.

```python
# Slide 46: (3,) * (2,) -- incompatible
x = np.array([1, 2, 3])
y = np.array([7, 8])
print(x.shape, y.shape)
try:
    x * y
except ValueError as err:
    print('ValueError:', err)
```

```python
# Slide 47: a scalar stretches to every element
x = np.array([1, 2, 3])
x * 4
```

```python
# Slide 48: (4, 3) + (3,) -> b stretches down the rows
a = np.array([[0, 0, 0], [10, 10, 10], [20, 20, 20], [30, 30, 30]])
b = np.array([1, 2, 3])
a + b
```

```python
# Slide 49: (4, 3) + (4,) fails: rightmost dims are 3 vs 4
try:
    a + np.array([1, 2, 3, 4])
except ValueError as err:
    print('ValueError:', err)
```

```python
# Fix: make it a column, shape (4, 1), so it stretches across the columns
a + np.array([1, 2, 3, 4]).reshape(4, 1)
```

```python
# Slide 50: check shapes without building big arrays
print(np.broadcast_shapes((256, 256, 3), (3,)))
print(np.broadcast_shapes((8, 1, 6, 1), (7, 1, 5)))
print(np.broadcast_shapes((15, 3, 5), (15, 1, 5)))
```
